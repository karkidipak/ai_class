```python
import numpy as np
from sklearn.datasets import make_regression
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.model_selection import train_test_split
from sklearn import tree

# Create data
X, y = make_regression(
    n_samples=100,
    n_features=1,
    noise=0.1,
    random_state=42
)

# Split data
x_train, x_test, y_train, y_test = train_test_split(
    X, y, test_size=0.5, random_state=42
)

# Create and train model
regr = tree.DecisionTreeRegressor()
regr.fit(x_train, y_train)

# Make predictions
predictions = regr.predict(x_test)

# Show predictions
print("Predictions:", predictions)
print("Actual:", y_test)

# Show first 5 samples
print("\nActual vs Predicted (top 5 samples):")

for i in range(5):
    print(
        f"x_test[{i}]: {x_test[i]} | "
        f"actual y: {y_test[i]:.2f} | "
        f"predicted y: {predictions[i]:.2f}"
    )

# Model evaluation
mse = round(mean_squared_error(y_test, predictions),2)
r2 = round(r2_score(y_test, predictions),4)

print("\nModel evaluation:")
print("MSE (Mean Squared Error):", mse)
print("R²:", r2)
```

```python
import numpy as np

# Data: 1D array with 3 elements
arr = np.array([1, 2, 3])
print(arr.shape)  # (3,) -> 1D array


# 1. Place np.newaxis at the row position -> (1, 3) 2D horizontal row vector
row_vec = arr[np.newaxis, :]
print(row_vec.shape)  # (1, 3)
print(row_vec)        # [[1, 2, 3]]


# 2. Place np.newaxis at the column position -> (3, 1) 2D vertical column vector
col_vec = arr[:, np.newaxis]
print(col_vec.shape)  # (3, 1)
print(col_vec)        # [[1],
                               #  [2],
                               #  [3]]
```

```python
import numpy as np
from sklearn.tree import DecisionTreeRegressor

# 1. Generate dataset
np.random.seed(42)
X = np.sort(5 * np.random.rand(100, 1), axis=0)
y = np.sin(X).ravel() + np.random.normal(0, 0.1, X.shape[0])

# 2. Train DecisionTreeRegressor models
regr_1 = DecisionTreeRegressor(max_depth=2)
regr_2 = DecisionTreeRegressor(max_depth=7)

regr_1.fit(X, y)
regr_2.fit(X, y)

# 3. Create test dataset X and predict
X_test = np.arange(0.0, 5.0, 0.01)[:, np.newaxis]

y_1 = regr_1.predict(X_test)
y_2 = regr_2.predict(X_test)
print(X_test[0:10])
print(y_1[0:10])
print("============================")
print(X_test[0:10])
print(y_2[0:10])
```
