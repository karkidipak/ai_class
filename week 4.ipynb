```python
import numpy as np
from sklearn.utils import Bunch
from sklearn import tree

x = np.random.randint(0, 10, size=(5, 3))
y = np.random.randint(7, 17, size=(5, 3))
z = np.random.randint(15, 24, size=(5, 3))


t1=np.zeros((5,))
t2=np.ones((5,))
t3=(np.ones((5,))+1)
data_bunch = Bunch(
    data=np.vstack((x,y,z)),
    data_names=["f1", "f2", "f3"],  # optional: column names
    description="15x3 random matrix",
    target=np.vstack((t1,t2,t3)).flatten().reshape(-1,1)
)
test= [0,5,15-1]
#testing data
train_data = np.delete(data_bunch.data, test,axis=0)
train_target = np.delete(data_bunch.target, test)

#testing data
test_data = data_bunch.data[test]
test_target = data_bunch.target[test]

#creating a decision tree
clf = tree.DecisionTreeClassifier()
clf.fit(train_data, train_target)

#output
print(test_target)
print(clf.predict([[5,9,10]]))

```

```python
import numpy as np
from sklearn.utils import Bunch
from sklearn import tree

x = np.random.randint(0, 15, size=(5, 3))
y = np.random.randint(10, 25, size=(5, 3))

t1=np.zeros((5,))
t2=np.ones((5,))

data_bunch = Bunch(
    data=np.vstack((x,y)),
    data_names=["f1", "f2"],  # optional: column names
    description="15x3 random matrix",
    target=np.vstack((t1,t2)).flatten()
                     )
text_idx=[0,5]
#testing data
train_data = np.delete(data_bunch.data, text_idx,axis=0)
train_target = np.delete(data_bunch.target, text_idx)

#testing data
test_data = data_bunch.data[text_idx]
test_target = data_bunch.target[text_idx]

#creating a decision tree
clf = tree.DecisionTreeClassifier()
clf.fit(train_data, train_target)

#output
print(test_target)
print(clf.predict(test_data))


```
