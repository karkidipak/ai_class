```python
import numpy as np
from sklearn.utils import Bunch
from sklearn import tree

x = np.random.randint(0, 15, size=(10, 3))
y = np.random.randint(10, 25, size=(10, 3))

t1=np.zeros((10,))
t2=np.ones((10,))

data_bunch = Bunch(
    data=np.vstack((x,y)),
    data_names=["f1", "f2","f3"],  # optional: column names
    description="20x3 random matrix",
    target=np.vstack((t1,t2)).flatten()
                     )
text_idx=[1,11]
#testing data
train_data = np.delete(data_bunch.data, text_idx,axis=0)
train_target = np.delete(data_bunch.target, text_idx)

#testing data
test_data = data_bunch.data[text_idx]
test_target = data_bunch.target[text_idx]

#creating a decision tree
clf = tree.DecisionTreeClassifier()
clf.fit(train_data, train_target)

#output
print(test_target)

print(clf.predict([[2,9,15],[10,19,25]]))

```

```python
from sklearn import tree
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
iris = load_iris()
features=iris.data
labels= iris.target
clf = tree.DecisionTreeClassifier()
clf = clf.fit(features, labels)
print(clf.predict([[5.1, 3.5, 1.4, 0.2]]))
plt.figure(figsize=(20,5))
tree.plot_tree(
    clf,feature_names=iris.feature_names, class_names=iris.target_names,filled=True
)

plt.savefig("../iris_tree.pdf", format="pdf")
plt.show()
```

```python
import numpy as np
from sklearn.datasets import load_iris
from sklearn import tree
iris = load_iris()
test= [0,50,100]
#testing data
train_data = np.delete(iris.data, test, axis=0)
train_target = np.delete(iris.target, test)

#testing data
test_data = iris.data[test]
test_target = iris.target[test]

#creating a decision tree
clf = tree.DecisionTreeClassifier()
clf.fit(train_data, train_target)

#output
print(test_target)
print(clf.predict(test_data))
```

```python

from sklearn import tree
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
iris = load_iris()
x = iris.data
y = iris.target
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2)

clf = tree.DecisionTreeClassifier()
clf = clf.fit(x_train, y_train)
predictions = clf.predict(x_test)
print(predictions)
print(y_test)

plt.savefig("../decision_tree.pdf", format="pdf", bbox_inches="tight")
plt.show()
```
